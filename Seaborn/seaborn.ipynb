# Seaborn

```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {"Days":[1,2,3,4,5],
        "NOP":[50,40,60,30,44]}
df = pd.DataFrame(data)
print(df)

sns.lineplot(data=data,x="Days",y="NOP")
plt.show()
```

```python
data = pd.read_csv("../pandas/sales.csv")
color = ["red", "blue", "green", "orange", "purple", "pink", "brown"]
sns.lineplot(data = data, x="unit_price",y="total_amount",hue="quantity",palette = color)
plt.show()
```

```python
data = sns.load_dataset("tips")
print(data)
sns.barplot(data = data , x='day',y="tip",hue="sex",palette = "viridis")
plt.plot()
```

```python
data = sns.load_dataset('titanic')
print(data)
sns.histplot(data = data,x = "age",hue = "who",kde=True)
plt.show()
```

```python
data = sns.load_dataset("tips")
print(data)
sns.scatterplot(data = data, x = "total_bill",y="tip",hue = "day",size="size",marker="o",palette="viridis")
```

```python
data = sns.load_dataset("tips")
gp = data.groupby("day").agg({"tip":"mean"})
sns.heatmap(gp)
plt.show()
```

```python
data = sns.load_dataset("tips")
sns.countplot(data = data,x="day")
plt.show()
```

```python
data = sns.load_dataset("tips")
sns.violinplot(data = data ,x="total_bill")
```

```python
data = sns.load_dataset("tips",)
sns.pairplot(data,hue = "day", diag_kind="hist")
plt.show()
```

```python
data = sns.load_dataset("tips")
sns.stripplot(data = data,x="day",y="total_bill",hue="sex",dodge=True)
plt.show()
```

```python
sns.boxplot(data = data, x="sex",y="tip",orient="vertical")
plt.show()
```

```python
# default is scatter plot
# catplot stands for categorical plot
sns.catplot(x="day",y="tip",data=data,hue="sex",kind="violin")
plt.show()
```

```python
data = sns.load_dataset("exercise")
sns.set_style(style="ticks")
sns.barplot(x = "time", y="pulse",data=data)
plt.show()
```

```python
sns.palplot(sns.color_palette())
plt.show()
```

```python
#multiple Plot
data = sns.load_dataset("tips")
a = sns.FacetGrid(data,col="time",height=2,hue="sex")
a.map(sns.scatterplot,"day","tip")
plt.show()      
```

```python
# relational Plot
sns.relplot(data = data,x="tip",y="total_bill",hue="sex")
plt.show()
```

```python
# Swarm Plot
sns.stripplot(data=data,x="day",y="total_bill",jitter=0.3,hue="sex",dodge = True)
plt.show()
```

```python
# KDE Plot

data = sns.load_dataset("tips")
sns.kdeplot(data = data,x="total_bill",hue ="day")
plt.show()
```
