# Box Plot

```python
l = [25,46,57,25,45,13,29,39,46,99,22,12,14,16,46]

plt.boxplot(l)
plt.show()
```

#### Box Plot: Shows the data’s spread, center, and outliers.
- Box Plot Formula:

- Median (Q2) = middle value
- Q1 = 25th percentile
- Q3 = 75th percentile
- IQR = Q3 − Q1
- Lower Bound = Q1 − 1.5 × IQR
- Upper Bound = Q3 + 1.5 × IQR

#### IQR (Interquartile Range): The range between Q1 and Q3.

- Formula: IQR = Q3 − Q1




```python
import matplotlib.pyplot as plt
import numpy as np

data = [10, 12, 15, 18, 20, 22, 25, 30, 35, 40]

q1 = np.percentile(data, 25)
median = np.percentile(data, 50)
q3 = np.percentile(data, 75)
iqr = q3 - q1

fig, ax = plt.subplots()

ax.boxplot(data, vert=False)

ax.text(q1, 1.05, "Q1", ha="center")
ax.text(median, 1.05, "Median", ha="center")
ax.text(q3, 1.05, "Q3", ha="center")

ax.annotate(
    f"IQR = {iqr:.2f}",
    xy=((q1 + q3) / 2, 0.9),
    ha="center"
)

plt.show()
```

# Stack Plot

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5, 6, 7]

NOP1 = [5, 10, 30, 20, 35, 60, 80]
NOP2 = [10, 20, 30, 25, 40, 50, 90]
NOP3 = [8, 30, 50, 60, 70, 90, 100]

plt.stackplot(
    days,
    NOP1, NOP2, NOP3,
    colors=["hotpink", "steelblue", "coral"],
    labels=["week1", "week2", "week3"]
)

plt.legend()
plt.show()
```

```python
grouped_by = df.groupby("product")[["quantity", "unit_price", "total_amount"]].mean()

print(grouped_by)

plt.stackplot(
    grouped_by.index,
    grouped_by["quantity"],
    grouped_by["unit_price"],
    grouped_by["total_amount"],
    labels=["Quantity", "Unit Price", "Total Amount"]
)

plt.legend()
plt.show()
```

# Sub Plot

```python
x = [1,2,3,4,5]
y = [45,34,56,23,45]

plt.subplot(1, 2, 1)
plt.title("Age")
plt.plot(x, y)

x = [5,6,7,8,9]
y = [20,40,30,50,60]

plt.subplot(1, 2, 2)
plt.title("No. Of Family")
plt.plot(x, y)
plt.suptitle("Population")
plt.show()
```

```python
x = [1,2,3,4,5]
y = [45,34,56,23,45]

plt.subplot(1, 2, 1)
plt.title("Age")
plt.step(x, y)

x = [5,6,7,8,9]
y = [20,40,30,50,60]

plt.subplot(1, 2, 2)
plt.title("No. Of Family")
plt.scatter(x, y)
plt.suptitle("Population")
#plt.savefig(graph.png)
plt.show()
```
