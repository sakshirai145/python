```python
data1.loc[data1["total_amount"] == 0, "SaleStatus"] = "No Sale"
data1.loc[data1["total_amount"] > 0, "SaleStatus"] = "Sale"

print(data1.head(10))
```

```python
data1.loc[data1["total_amount"] <= 3000, "SaleStatus"] = "Low Sale"
data1.loc[(data1["total_amount"] > 5000) & (data1["total_amount"] <= 10000), "SaleStatus"] = "Medium Sale"
data1.loc[data1["total_amount"] > 10000000, "SaleStatus"] = "High Sale"

print(data1.head(10))
```

```python
data1 = {
    "employee_id": [101, 102, 106, 108, 105],
    "age": [25, 28, 32, 27, 30],
    "name": ["Sakshi", "Rahul", "Priya", "Aman", "Neha"]
}
data2 = {
    "employee_id": [101, 102, 103, 104, 105],
    "salary": [35000, 42000, 55000, 40000, 48000]
}

df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
print(df1)
print()
print(df2)
```

```python
data = {
    "Fruit": ["Apple", "Banana", "Mango", "Orange", "Grapes"],
    "Quantity": [10, 20, 15, 12, 8],
    "Price": [100, 60, 120, 80, 90]
}

df1 = pd.DataFrame(data)

print(df1)
df2= df1.copy()
df2.loc[0,"Price"] = 120
df2.loc[1,"Price"] = 175
df2.loc[3,"Price"] = 30
df2.loc[0,"Quantity"] = 12
df2.loc[1,"Quantity"] = 15
df2.loc[3,"Quantity"] = 5
print(df2)
print(df1.compare(df2 , keep_shape= True))
```

```python
data = {
    "Keys": [101, 102, 103, 104, 105],
    "Names": ["Sakshi", "Rahul", "Priya", "Aman", "Neha"],
    "Houses": ["A-12", "B-15", "C-15", "D-08", "E-21"],
    "Grades":["3rd", "8th","9th","8th","4th"]
}

df = pd.DataFrame(data)

print(df)
print(df.pivot(index="Keys",columns="Names",values=["Houses","Grades"]))
```

```python
data = {
    "Names": ["Sakshi", "Rahul", "Priya", "Aman", "Neha"],
    "Section": ["A-12", "B-15", "C-15", "D-08", "E-21"],
    "Grades":["3rd", "8th","9th","8th","4th"]
}

df = pd.DataFrame(data)

print(df)
print(pd.melt(df, id_vars=["Names"],value_vars=["Section","Grades"]))
```
