# Why Hashing?
- It search the item in O(1) time complexity.
- It provide fast searching

### Searching
- Linear Search :-           1. compare item one by one.
-                            2. Time complexity O(n).

- Binary Search :-           1. Searches item by halving a sorted array.
-                            2. mid = low + (high - low) // 2
-                            3.  Time Complexity O(log n).


## Hashing
#### Formula: 
- Hash Index = Key ÷ Array Size → Remainder
- It does not depend on the length of the list.

#### Disadvantage:
- Space inefficiency due to sparsity(Wasted empty space).

#### Hashing Function:
- Hashing Function: Converts key into index.
- Reduces wasted space in hashing.
- solve collision 

#### Collision in Hashing
- Collision: Two keys get the same index.
#### Types:-
- Open Addressing: Find another empty index in the same table after collision.
- Closed Addressing (Chaining): Store multiple keys at the same index using a list/chain.

#### Chaining has time complexity O(n).
- To overcome we use Rehashing and Tree.
- Rehashing:  Creates a bigger table and hashes again.


##### Types of open addressing : 
##### Linear Probing:
-  Checks the next index for an empty spot.
-  New Index = (Hash Index + 1) % Table Size
##### Disadvantage:
- Causes clustering and slows searching.
##### Quadratic Probing:
- Checks positions using square jumps.
- New Index = (Hash Index + i²) % Table Size
- General Formula: New Index = (h(k) + i²) % m
##### Disadvantage:
- Can cause secondary clustering.
