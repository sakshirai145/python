# Searching
### searching are of two type
- Linear Search
- Binary Search


```python
def linear_search(arr, item):
    for i in range(len(arr)):
        if arr[i] == item:
            return i
    return -1    
```

```python
arr = [11, 23, 33, 44, 55]
print(linear_search(arr, 55))
```

```python
arr = [11, 55, 33 , 44, 23, 55]
print(linear_search(arr, 55))
```

### Time Complexity for Linear Search
- O(n)
- No sorting required

```python
def binary_search(arr, low, high, item):
    print("low= ", low, "high= ", high, end=' ')
    if low <= high:
        mid = (low + high) // 2
        print("mid value is ", arr[mid])
        if arr[mid] == item:
            return mid
        elif arr[mid] > item:
            return binary_search(arr, low , mid-1, item)
        else:
            return binary_search(arr, mid+1, high, item)
    else:
        return -1
```

```python
arr = [12, 23, 34, 45, 56, 67]
print(binary_search(arr, 0, len(arr)-1, 45))
```
