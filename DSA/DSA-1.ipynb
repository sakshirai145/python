# Array

- linear data structure.
- used to store data in contiguous memory loctation.

### TimeComplexity of write data in Dynamic Array 
- o(n)
- for normal array it is 0(1)

```python
class Meralist:

    def __init__(self):
        self.size = 1                  # Initial capacity
        self.n = 0                     # Current number of elements
        self.A = self.__make_array(self.size)  # Initialize underlying array


    def __make_array(self, capacity):
        return (capacity * ctypes.py_object)()  # Create fixed-size C-type array


    def __len__(self):
        return self.n                   # Length tracking


    def __resize(self, new_capacity):
        B = self.__make_array(new_capacity)  # Dynamic resizing: create new array

        for i in range(self.n):
            B[i] = self.A[i]             # Element copying

        self.A = B
        self.size = new_capacity        # Update capacity


    def __str__(self):
        result = ''

        for i in range(self.n):
            result += str(self.A[i]) + ','  # String traversal

        return '[' + result[:-1] + ']'     # Convert array to string


    def __getitem__(self, index):
        if 0 <= index < self.n:
            return self.A[index]           # Index-based access

        else:
            return 'IndexError -Index out of bounds'


    def append(self, item):
        if self.n == self.size:
            self.__resize(2 * self.size)   # Dynamic resizing: double capacity

        self.A[self.n] = item              # Append at end
        self.n = self.n + 1                # Update element count


    def pop(self):
        if self.n == 0:
            return 'Empty list'

        print(self.A[self.n - 1])          # Remove from end: LIFO
        self.n = self.n - 1                # Decrease element count


    def clear(self):
        self.n = 0                         # Logical deletion: reset element count
        self.size = 1                      # Reset capacity


    def find(self, item):
        for i in range(self.n):
            if self.A[i] == item:
                return i                   # Linear search: return matching index

        return 'ValueError - Item not found in the list'


    def insert(self, pos, item):
        if self.n == self.size:
            self.__resize(self.size * 2)   # Dynamic resizing: double capacity

        for i in range(self.n, pos, -1):
            self.A[i] = self.A[i - 1]      # Right shift: make insertion space

        self.A[pos] = item                 # Insert at given position
        self.n = self.n + 1                # Update element count


    def remove(self, item):
        pos = self.find(item)               # Linear search: find item position

        if type(pos) == int:
            self.__delitem__(pos)          # Delete by index

        else:
            return pos                     # Item-not-found handling


    def __delitem__(self, pos):
        if 0 <= pos < self.n:
            for i in range(pos, self.n - 1):
                self.A[i] = self.A[i + 1]  # Left shift: fill deleted position

            self.n = self.n - 1            # Update element count
```

# Dynamic Array

1. size = total available capacity
2. n    = number of actual elements
3. A    = actual storage
