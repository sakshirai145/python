```python
class Meralist:

    def __init__(self):
        self.size = 1                  # Initial capacity of the array
        self.n = 0                     # Number of elements currently stored
        self.A = self.__make_array(self.size)  # Create C type  the actual array


    def __make_array(self, capacity):
        return (capacity * ctypes.py_object)()  # Create a C type array (static, referential) with size capacity.


    def __len__(self):
        return self.n

    def __resize(self, new_capacity):
        # create a new array with the new capacity
        B = self.__make_array(new_capacity)
        # copy the elements from the old array(A) to the new array(B)
        for i in range(self.n):
            B[i] = self.A[i]
        self.A = B 
        self.size = new_capacity  # update the size of the array to the new capacity  


    def __str__(self):
        result = ''
        for i in range(self.n):
            result +=  str(self.A[i]) + ','
        return '[' + result[:-1] + ']'

    def __getitem__(self, index):
        if 0<= index < self.n:
            return self.A[index]
        else:
            return 'IndexError -Index out of bounds'


    def append(self,item):
        if self.n == self.size:
            # resize 
            self.__resize(2 * self.size)  # copy the element to diffrent list if the size of the array if it is full and double the size of the array.

            # append the item to the end of the array
        self.A[self.n] = item
        self.n = self.n + 1 


    def pop(self):
        if self.n == 0:
            return 'Empty list'
        print(self.A[self.n-1])
        self.n = self.n - 1 

    def clear(self):
        self.n=0 
        self.size =1            
```

# Dynamic Array

1. size = total available capacity
2. n    = number of actual elements
3. A    = actual storage
