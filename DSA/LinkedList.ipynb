```python
class Node:
    def __init__(self,value):
        self.data = value
        self.next = None
        
```

```python
class LinkedList:
    def __init__(self):
        # Empty Linked List
        self.head = None
        self.n = 0 # noumeber of nodes in linked list

    def __len__(self):
        return self.n   
    def insert_head(self,value):
        #new node to be inserted
        new_node = Node(value)
        #point new node to head
        new_node.next = self.head
        #update head to new node
        self.head = new_node 
        # increment the number of nodes
        self.n = self.n + 1

    def traverse(self):
        curr = self.head
        result = ''
        while curr != None:
            result += str(curr.data) + ' -> '
            curr = curr.next
        return result[:-4]  # remove the last arrow

    def append(self, value):
        new_node = Node(value)
        if self.head == None:
            self.head = new_node
            self.n = self.n +1
            return
        curr = self.head
        while curr.next != None:
            curr = curr.next
        curr.next = new_node 
        self.n = self.n + 1

    def insert_after(self,after, value):
        new_node = Node(value)
        curr = self.head
        while curr != None:
            if curr.data == after:
                break
            curr = curr.next
        if curr != None:
            new_node.next = curr.next
            curr.next = new_node
            self.n = self.n + 1
        else:
            return 'Item not found'     

    def clear(self):
        self.head = None
        self.n = 0 

    def delete_head(self):
        if self.head == None:
            return 'LinkedList is empty'
        self.head = self.head.next
        self.n = self.n -1
        if self.head == None:
            return 'LinkedList is empty'   


    def tailpop(self):
        if self.head == None:
            return 'LinkedList is empty'
        if self.head.next == None:
            self.head = None
            self.n -=1
            return
        curr = self.head
        while curr.next.next != None:
            curr = curr.next
        curr.next = None
        self.n = self.n -1   

    def remove(self, value):
        if self.head == None:
            return 'LinkedList is empty'
        if self.head.data == value:
            print('Node deleted')
            return self.delete_head()
             

        curr = self.head
        while curr.next != None:
            if curr.next.data == value:
                curr.next = curr.next.next
                self.n -= 1
                print('Node deleted')
                return
            curr = curr.next
        return 'Item not found'   


    def search(self, value):
       curr = self.head
       pos = 0

       while curr != None:
           if curr.data == value:
               print('found')
               return pos
           curr = curr.next
           pos = pos + 1
       return 'Not Found'  


    def __getitem__(self, index):
        curr = self.head
        pos = 0

        while curr != None:
            if pos == index:
                return curr.data
            curr = curr.next
            pos = pos +1 
        return 'Index Not found'  

    


    
```
