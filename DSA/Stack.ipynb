# Stack

```python
class Node:
    def __init__(self,value):
        self.data = value
        self.next = None
        
```

```python
class stack:
    def __init__(self):
        self.top = None

    def isempty(self):
        return self.top == None
    def push(self,value):
        new_node = Node(value)
        new_node.next = self.top
        self.top = new_node
    def traverse(self):
        temp = self.top
        while temp != None:
            print(temp.data) 
            temp = temp.next  

    def peek(self):
        if(self.isempty()):
            return "stack is Empty"
        else:
            return self.top.data

    def pop(self):
        if(self.isempty()):
            return "stack is Empty" 
        else:
            value = self.top.data
            self.top = self.top.next
            return value

                 

    
```

```python
def reverse_string(text):
    s= stack()
    for i in text:
        s.push(i)
    res =""
    while(not s.isempty()):
        res = res + s.pop()
    return res       
```

```python
def find_celebrity(L):
    s= stack()
    for i in range(len(L)):
        s.push(i)
    while True:
        i = s.pop()
        if s.isempty():
            celeb = i
            break
        j = s.pop()
        if L[i][j] == 0:
            s.push(i)
        else:   
            s.push(j)

    for i in range(len(L)):
        if i != celeb:
            if L[i][celeb]== 0  or L[celeb][i] == 1:
                print("No one is celebrity")
                return
    print("The celebrity is", celeb)       
```

```python
#Stack array implementation
class stack:
    def __init__(self,size):
        self.size = size
        self.stack = [None] * self.size
        self.top = -1
    def push(self,value):
        if self.top == self.size -1:
            return "Overflow"
        else:
            self.top+=1
            self.stack[self.top]=value
    def pop(self):
        if self.top == -1:
            return "Empty"
        else:
            data = self.stack[self.top]
            self.top -=1
            print(data)

    def traverse(self):
        for i in range(self.top +1):
            print(self.stack[i], end='')         
                     
```
