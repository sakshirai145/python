```python
C3 = np.array([30,80,90,40,50,60])
print(np.sum(C3))
print(np.min(C3))
print(np.max(C3))
print(np.size(C3))
print(np.mean(C3))
print(np.cumsum(C3))
print(np.cumprod(C3))
```

```python
p = [100,150,199,200,250,130]
q = [10,50,30,40,30,10]
price = np.array(p)
quantity = np.array(q)

print(price,"\n",quantity)
print()
tot = np.cumprod([price, quantity], axis=0)

print(tot)
print(tot[1].sum())
```

```python
import statistics as stats
baked_food = [200,300,150,130,200,280,170,188]
a = np.array(baked_food)
print(np.mean(baked_food))
print(np.median(baked_food))
print(stats.mode(baked_food))
print(np.std(baked_food))
print(np.var(baked_food))

```
